```python
## 1)WAP to reverse a string
str = 'Python'
s1 = str[::-1]
print(s1)
```

```python
## 2) WAP to reverse a string using for loop
str = 'Python'
s1 = str.split()
s2 = []
for word in s1:
  s2.append(word[::-1])
s3 = ''.join(s2)
print(s3)
```

```python
## 3) WAP to reverse a string using while loop
str = 'Python'
s1 = ''
i = len(str)-1
while i>=0:
  s1 = s1+str[i]
  i=i-1
print(s1)
```

```python
## 4)WAP to reverse only words in a string
s = 'Python is a language'
s1 = s.split()[::-1]
s2 = ' '.join(s1)
print(s2)
```

```python
## 5)WAP to reverse all the letters in a string
s = 'Python is a language'
s1 = s.split()
s2 = []
for word in s1:
  s2.append(word[::-1])
s3 = ' '.join(s2)
print(s3)
```

```python
## 6)WAP to find even odd numbers in a list
lst = [15,88,73,97,49,51,90,53,91,63,80,88]
even = []
odd = []
for i in range(len(lst)):
  if lst[i]%2 == 0:
    even.append(lst[i])
  else:
    odd.append(lst[i])
print('Even numbers are:',even)
print('Odd numbers are:',odd)
```

```python
## 7)WAP to find even odd numbers in a list using list comprehension
lst = [15,88,73,97,49,51,90,53,91,63,80,88]
even = [i for i in lst if i%2==0]
odd = [i for i in lst if i%2!=0]
print('Even numbers are:',even)
print('Odd numbers are:',odd)
```

```python
## 8)WAP to find even odd numbers in a list using lambda
lst = [15,88,73,97,49,51,90,53,91,63,80,88]
even = list(filter(lambda x:x%2==0,lst))
odd = list(filter(lambda x:x%2!=0,lst))
print('Even numbers are:',even)
print('Odd numbers are:',odd)
```

```python
## 10) WAP to find max of three numbers
def maximum (x,y,z):
  if x>=y and x>=z:
    return x
  elif y>=x and y>=z:
    return y
  else:
    return z
print(maximum(10,10,9))
```

```python
## 11)WAP to find max number in a list
lst = [22,77,6,94,49,60,17,52]
max = lst[0]
for i in range(len(lst)):
  if lst[i]>max:
    max=lst[i]
print(max)
```

```python
## 12)WAP to find max number in a list using lambda
lst = [22,77,6,94,49,60,17,52]
from functools import reduce
max = reduce(lambda x,y :x if x>y else y,lst)
print(max)
```
